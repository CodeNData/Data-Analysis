## Amount by Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Amount"], bins=30, kde=True)
plt.title("Transaction Amount Distribution")
plt.show()
```

## Amount by City

```python
plt.figure(figsize=(10,6))
sns.barplot(x="City", y="Amount", data=df, estimator="mean", errorbar=None)
plt.title("Average Transaction Amount by City")
plt.xticks(rotation=45)
plt.show()
```

## Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["CustomerAge"], bins=20, kde=True)
plt.title("Customer Age Distribution")
plt.show()
```

## Transactions Over Time

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"])
txn_per_day = df.groupby("TransactionDate").size()

plt.figure(figsize=(12,6))
txn_per_day.plot(kind="line")
plt.title("Number of Transactions Over Time")
plt.ylabel("Count")
plt.show()
```

## Transactions by hour

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Hour", data=df)
plt.title("Transactions by Hour of Day")
plt.show()
```

## Bank (Recieved Vs Sent)

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Hour", data=df)
plt.title("Transactions by Hour of Day")
plt.show()
```

## Status (Success/Failed)

```python
plt.figure(figsize=(6,6))
df["Status"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Transaction Status Distribution")
plt.ylabel("")
plt.show()
```

## Gender-based spend

```python
plt.figure(figsize=(8,6))
sns.barplot(x="Gender", y="Amount", data=df, estimator="mean", errorbar=None)
plt.title("Average Transaction Amount by Gender")
plt.show()
```

## Transaction Type

```python
plt.figure(figsize=(6,6))
df["TransactionType"].value_counts().plot(kind="bar", color="skyblue")
plt.title("Transaction Type Distribution")
plt.show()
```

## DeviceType usage

```python
plt.figure(figsize=(8,6))
sns.countplot(x="DeviceType", data=df)
plt.title("Transactions by Device Type")
plt.show()
```

## PaymentMethod

```python
plt.figure(figsize=(10,6))
sns.countplot(y="PaymentMethod", data=df, order=df["PaymentMethod"].value_counts().index)
plt.title("Transactions by Payment Method")
plt.show()
```

## Purpose of transaction

```python
plt.figure(figsize=(10,6))
sns.barplot(x="Purpose", y="Amount", data=df, estimator="sum", errorbar=None)
plt.title("Total Spending by Purpose")
plt.xticks(rotation=45)
plt.show()
```

## Currency split

```python
plt.figure(figsize=(6,6))
df["Currency"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Transactions by Currency")
plt.ylabel("")
plt.show()
```
