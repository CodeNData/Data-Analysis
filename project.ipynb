```python
import numpy as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
project=pd.read_csv("./sales_data.csv",encoding='latin1')
project
```

```python
project.info()
```

```python
#checking numebr of rows and colomns 
project.shape
```

```python
# checking for missing values
project.isnull()
```

```python
project.isnull().sum()
```

```python
project.columns
```

```python
# formaing
project.columns=project.columns.str.replace(' ',"_")
```

```python
project.columns
```

```python
# filling 12 null values with avg amount of column.
project['Amount']=project['Amount'].fillna(project['Amount'].mean())
```

```python
# checking is amount column creating null values or not 
project.duplicated().sum()
```

```python
# removing duplicate rows from the dataset 
project.drop_duplicates(inplace=True)
```

```python
# changing the value present in gender column
project['Gender']=project['Gender'].map({'M':'Male','F':'Female'})
```

```python
project['Marital_Status']=project['Marital_Status'].map({1:'Married',0:'Single'})
```

```python
project.head()
```

```python
project.drop('unnamed1',axis=1,inplace=True)
```

```python
project.head()
```

```python
project.drop('Status',axis=1,inplace=True)
```

```python
project.head()
```

# data visualization

```python
def label(lab):
    for i in lab.containers:
        lab.bar_label(i)
```

```python

a=sns.countplot(x='Gender',data=project,palette='hls',hue='Gender')
label(a)
```

```python
g= project.groupby('Gender')['Amount'].sum()
plt.pie(g,labels=g.index,autopct='%1.2f%%',shadow=True)
```

```python
sns.barplot(x='Gender',y='Amount',data=project,palette='hls',hue='Gender')
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(15,7))
a= sns.countplot(x='Gender',data=project,palette='hls',hue='Gender',ax=axes[0])
label(a)

g_g =  project.groupby('Gender')['Amount'].sum()
plt.pie(g_g,labels=['Female','Male'],
        colors=["#5ca8a9","#4eb151"],
        explode=(0,0.1),shadow=True,autopct='%1.2f%%',startangle=90)
```

```python
fig_1,axes = plt.subplots(nrows=1,ncols=2,figsize=(25,10))
sns.barplot(x='Age_Group',y='Amount',data=project,hue='Age_Group',ax=axes[0],errorbar=None,
            estimator='sum',palette='Set2')
sns.barplot(x='State',y='Amount',data=project,hue='State',ax=axes[1],errorbar=None,
            estimator='sum',palette='Set2')
plt.xticks(rotation= 90)
plt.show()
```

```python
fig_2,axes = plt.subplots(nrows=1,ncols=2,figsize=(25,10))
sns.barplot(x='Gender',y='Amount',data=project,hue='Gender',ax=axes[0],errorbar=None,
            estimator='sum',palette='Set2')
sns.barplot(x='Occupation',y='Amount',data=project,hue='Occupation',ax=axes[1],errorbar=None,
            estimator='sum',palette='Set2')

plt.xticks(rotation= 90)
plt.show()
```

```python
fig_2,axes = plt.subplots(nrows=1,ncols=2,figsize=(25,25))
sns.barplot(y='Product_Category',x='Amount',data=project,hue='Product_Category',ax=axes[0],errorbar=None,
            estimator='sum',palette='Set2')
plt.xticks(rotation= 60)
sns.barplot(x='Marital_Status',y='Amount',data=project,hue='Marital_Status',ax=axes[1],errorbar=None,
            estimator='sum',palette='Set2')
plt.xticks(rotation= 90)
plt.show()
```
